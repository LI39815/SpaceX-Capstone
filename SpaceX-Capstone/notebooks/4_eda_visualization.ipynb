# 4. Exploratory Data Analysis — Visualisation

Explore launch patterns using matplotlib and seaborn.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style='whitegrid', palette='muted')
```

```python
df = pd.read_csv('../data/dataset_part_2.csv')
print('Shape:', df.shape)
df.head()
```

## 4.1 Flight Number vs. Launch Site

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='FlightNumber', y='LaunchSite',
                hue='Class', palette={0:'red',1:'green'}, s=60)
plt.title('Flight Number vs Launch Site')
plt.xlabel('Flight Number')
plt.ylabel('Launch Site')
plt.legend(title='Success', labels=['Failure','Success'])
plt.tight_layout()
plt.show()
```

## 4.2 Payload Mass vs. Launch Site

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='PayloadMass', y='LaunchSite',
                hue='Class', palette={0:'red',1:'green'}, s=60)
plt.title('Payload Mass vs Launch Site')
plt.xlabel('Payload Mass (kg)')
plt.ylabel('Launch Site')
plt.tight_layout()
plt.show()
```

## 4.3 Success Rate by Orbit

```python
orbit_success = df.groupby('Orbit')['Class'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,5))
orbit_success.plot(kind='bar', color='steelblue', edgecolor='white')
plt.title('Average Success Rate by Orbit Type')
plt.xlabel('Orbit')
plt.ylabel('Success Rate')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 4.4 Flight Number vs. Orbit Type

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='FlightNumber', y='Orbit',
                hue='Class', palette={0:'red',1:'green'}, s=60)
plt.title('Flight Number vs Orbit Type')
plt.tight_layout()
plt.show()
```

## 4.5 Payload Mass vs. Orbit Type

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='PayloadMass', y='Orbit',
                hue='Class', palette={0:'red',1:'green'}, s=60)
plt.title('Payload Mass vs Orbit Type')
plt.xlabel('Payload Mass (kg)')
plt.tight_layout()
plt.show()
```

## 4.6 Yearly Launch Success Trend

```python
# Parse year from Date column
df['Year'] = pd.to_datetime(df['Date']).dt.year

yearly = df.groupby('Year')['Class'].mean().reset_index()
yearly.columns = ['Year','SuccessRate']

plt.figure(figsize=(10,5))
plt.plot(yearly['Year'], yearly['SuccessRate'], marker='o', linewidth=2.5, color='steelblue')
plt.fill_between(yearly['Year'], yearly['SuccessRate'], alpha=0.15, color='steelblue')
plt.title('Yearly Launch Success Rate')
plt.xlabel('Year')
plt.ylabel('Success Rate')
plt.ylim(0, 1.1)
plt.xticks(yearly['Year'], rotation=45)
plt.tight_layout()
plt.show()
```

## 4.7 Success Count by Launch Site

```python
site_counts = df.groupby(['LaunchSite','Class']).size().unstack(fill_value=0)
site_counts.columns = ['Failure','Success']

site_counts.plot(kind='bar', figsize=(9,5), color=['salmon','steelblue'], edgecolor='white')
plt.title('Launch Outcomes by Site')
plt.xlabel('Launch Site')
plt.ylabel('Count')
plt.xticks(rotation=15, ha='right')
plt.legend(title='Outcome')
plt.tight_layout()
plt.show()
```
