# 3. Data Wrangling

Clean data, engineer features, and create the binary **Class** label.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

## 3.1 Load Data

```python
df = pd.read_csv('../data/dataset_part_1.csv')
print('Shape:', df.shape)
df.head(5)
```

## 3.2 Explore Missing Values

```python
# Percentage of missing values per column
missing = df.isnull().sum() / len(df) * 100
missing = missing[missing > 0].sort_values(ascending=False)
print('Missing value percentages:\n')
print(missing)
```

```python
# Visualise
missing.plot(kind='barh', figsize=(8,4), color='steelblue')
plt.xlabel('Missing %')
plt.title('Missing Values by Column')
plt.tight_layout()
plt.show()
```

## 3.3 Explore Launch Outcomes

```python
# Count each unique outcome
print(df['Outcome'].value_counts())
```

```python
# Define bad outcomes (landing failure)
bad_outcomes = set(['False ASDS', 'None ASDS', 'None None',
                    'False Ocean', 'False RTLS', 'None RTLS',
                    'False Ocean', 'Controlled (ocean)'])

# Create binary Class label: 1 = success, 0 = failure
df['Class'] = df['Outcome'].apply(lambda x: 0 if x in bad_outcomes else 1)

print('Class distribution:')
print(df['Class'].value_counts())
print('\nOverall success rate: {:.2f}%'.format(df['Class'].mean()*100))
```

## 3.4 Feature Overview

```python
# Unique values in categorical columns
for col in ['BoosterVersion','Orbit','LaunchSite','LandingPad']:
    print(f'{col}: {df[col].nunique()} unique values')
    print(' ', df[col].unique(), '\n')
```

## 3.5 Save Wrangled Data

```python
df.to_csv('../data/dataset_part_2.csv', index=False)
print('Saved dataset_part_2.csv')
print('Final shape:', df.shape)
df[['FlightNumber','BoosterVersion','Orbit','LaunchSite','Outcome','Class']].head(10)
```
