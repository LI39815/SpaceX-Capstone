# 5. EDA with SQL

Use SQL queries via `sqlite3` to analyse the SpaceX dataset.

```python
import pandas as pd
import sqlite3

# Load CSV and store in SQLite in-memory DB
df = pd.read_csv('../data/dataset_part_2.csv')
con = sqlite3.connect('spacex.db')
df.to_sql('SPACEXTBL', con, if_exists='replace', index=False)
print('Table created. Shape:', df.shape)
```

## Q1 — Display the names of the unique launch sites

```python
pd.read_sql('SELECT DISTINCT "LaunchSite" FROM SPACEXTBL', con)
```

## Q2 — Display 5 records where launch sites begin with 'CCA'

```python
pd.read_sql('SELECT * FROM SPACEXTBL WHERE "LaunchSite" LIKE \'CCA%\' LIMIT 5', con)
```

## Q3 — Total payload mass carried by NASA (CRS)

```python
pd.read_sql('''
    SELECT SUM("PayloadMass") AS TotalPayload
    FROM SPACEXTBL
    WHERE "Customer" LIKE '%NASA%'
''', con)
```

## Q4 — Average payload mass by booster version F9 v1.1

```python
pd.read_sql('''
    SELECT AVG("PayloadMass") AS AvgPayload
    FROM SPACEXTBL
    WHERE "BoosterVersion" = 'F9 v1.1'
''', con)
```

## Q5 — Date of first successful ground landing

```python
pd.read_sql('''
    SELECT MIN("Date") AS FirstSuccessfulLanding
    FROM SPACEXTBL
    WHERE "Outcome" LIKE '%True RTLS%'
''', con)
```

## Q6 — Boosters that succeeded in drone ship landings with payload 4000–6000 kg

```python
pd.read_sql('''
    SELECT "BoosterVersion"
    FROM SPACEXTBL
    WHERE "Outcome" LIKE '%True ASDS%'
      AND "PayloadMass" BETWEEN 4000 AND 6000
''', con)
```

## Q7 — Total success and failure mission outcomes

```python
pd.read_sql('''
    SELECT "Class",
           COUNT(*) AS Count
    FROM SPACEXTBL
    GROUP BY "Class"
''', con)
```

## Q8 — Boosters with max payload mass

```python
pd.read_sql('''
    SELECT "BoosterVersion", "PayloadMass"
    FROM SPACEXTBL
    WHERE "PayloadMass" = (SELECT MAX("PayloadMass") FROM SPACEXTBL)
''', con)
```

## Q9 — Monthly failure records in 2015

```python
pd.read_sql('''
    SELECT strftime('%m', "Date") AS Month,
           "LaunchSite",
           "BoosterVersion",
           "Outcome"
    FROM SPACEXTBL
    WHERE strftime('%Y', "Date") = '2015'
      AND "Class" = 0
''', con)
```

## Q10 — Rank landing outcomes between 2010-06-04 and 2017-03-20

```python
pd.read_sql('''
    SELECT "Outcome",
           COUNT(*) AS Count
    FROM SPACEXTBL
    WHERE "Date" BETWEEN '2010-06-04' AND '2017-03-20'
    GROUP BY "Outcome"
    ORDER BY Count DESC
''', con)
```

```python
con.close()
print('Done.')
```
